# Test/Train split

```python
# split the dataset
X=df1.drop(columns=['CONDITION'])
y=df1['CONDITION']
```

```python
from sklearn.model_selection import train_test_split

#Test/Train split using 70/30 split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
```

# Build Random Forest model

```python
#import library
from sklearn.ensemble import RandomForestClassifier

rf=RandomForestClassifier(
    n_estimators=400,  # number of trees is acceptable
    max_depth=5,    # must be regulated to prevent model memorization of data
    class_weight='balanced',  # to address class imbalance
    max_features='sqrt', # selects a subset of features and not all features 
    min_samples_leaf=2,  # prevents tiny leaves 
    random_state=42,
    n_jobs=-1            # No of jobs working concurrently
)
```

# Train the model

```python
rf.fit(X_train, y_train)
```

# Use trained model to make predictions

```python
y_pred=rf.predict(X_test)
```

# Evaluation

Model is evaluated against the following metrics:
1. Accuracy
2. Confusion Matrix
3. Classification report
4. Cross validation
5. Out-of-Bag score
6. Feature importance

```python
# 1. Check model accuracy
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)  # compares the target in the test split with the predicted target
```

Model predicts with a 0.86670 (86.67%) accuracy, even after addressing class imbalance

```python
# 2. Plot confusion matrix
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

sns.heatmap(cm, annot=True, fmt='d')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
# 3. Classification report
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
# 4. Cross validation
from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(rf, X_train, y_train, cv=5)
print(cv_scores.mean())
```

1. Since cv_score (0.8548) is approximately equal to test/accuracy score (0.8667), we have good generalisation!
2. The cv_score >> Test score could indicate several issues, such as data leakage

```python
# 5. Out-of-Bag score
rf_oob = RandomForestClassifier(
    n_estimators=400,
    oob_score=True,
    random_state=42
)

rf_oob.fit(X_train, y_train)

print(rf_oob.oob_score_)
```

1. OOB score approximates the cross-validation score. Another good indicator of model generalisation

```python
# 6. Feature Importance

importance_df1 = pd.DataFrame({
    'feature': X_train.columns,
    'importance': rf.feature_importances_
}).sort_values(by='importance', ascending=False)

print(importance_df1)
```

Provides deep insights on infrastructural and engineering interventions

1. The Asthma Concrete thickness (AC_THICKNESS) and Construction Number (CONSTRUCTION_NO) which measure pavement thickness and material quality emerge as the strongest predictor of pavement deterioration.
2. It correlates to pavement performance by indicating that the structural capacity at the time of construction is a more dominant driver of long-term deterioration than traffic or climate alone.
